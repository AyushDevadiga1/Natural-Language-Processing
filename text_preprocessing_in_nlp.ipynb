```python
# Now let's automate this process and try to get a large number of json in a iteration.
def fetch_json_movies(url,params):
  movies = []
  first_page = 1
  max_pages = 1
  try:
      params['page']=first_page
      response = requests.get(url,params=params)

      response.raise_for_status() # This triggers HTTPError if the response is not 200-299

      fetched_movies = response.json()

      max_pages = fetched_movies['total_pages']

      movies.extend(fetched_movies['results'])

  except requests.exceptions.HTTPError as err:
    print(f"Caught an error before sending request: {err}")
    return movies
  except requests.exceptions.RequestException as net_err:
    print(f"Page {first_page}: Network/Timeout error occurred: {net_err}")
    return movies

  for i in range(first_page+1,max_pages+1): # Temporary for now
    params["page"] = i
    try:
      response = requests.get(url,params=params)

      response.raise_for_status() # This triggers HTTPError if the response is not 200-299

      fetched_movies = response.json()
      movies.extend(fetched_movies['results'])

    except requests.exceptions.HTTPError as err:
      print(f"Page {i}: Bad status code received: {err}")
      break
    except requests.exceptions.RequestException as net_err:
      print(f"Page {i}: Network/Timeout error occurred: {net_err}")
      break
  return movies
```

```python
# Set url and parameters
url = "https://api.themoviedb.org/3/movie/top_rated"
params = {
    "api_key": TMDB_API_KEY, # The API_KEY hidden in colab secrets
    "language":"en-US",
    "page":1 # Dummy for now , replace by the iterator in loop
}
all_movies = fetch_json_movies(url,params)
```

```python
# Now there is another mappig endpoint we have to fetch :
import requests
import pandas as pd

# 1. Set up the exact URL and parameters (Remove page since it isn't needed here)
url = "https://api.themoviedb.org/3/genre/movie/list"
params = {
    "api_key": TMDB_API_KEY,
    "language": "en-US"
}

def fetch_genres(url, params):
    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()

        data = response.json()

        # Safely extract the 'genres' list (default to empty list if missing)
        return data.get('genres', [])

    except requests.exceptions.RequestException as err:
        print(f"Failed to fetch genres: {err}")
        return []

# 2. Fetch the data using the clean function
all_genres = fetch_genres(url, params)

# 3. Convert directly to a beautiful Pandas DataFrame
genres = pd.DataFrame(all_genres)
```

###  So now we have to clean the overview column moving forward:
- First we will conver the entire column text for each row into lowercase.
- We dont have unwanted urls,html tags ,emojis, or short forms judging from the dataset.
- Also since this data is from movies written carefully there is low chance that there are spelling mistakes.
- We are going to remove punctuation marks next.
- We are then going to remove stopwords.
- Then we tokenize the sentences in words
- Then we are gonna use either of stemming or lemmatization.

```python
import re
```

```python
def preprocess_overview(text):
    # 1. Remove URLs
    text = re.sub(r"https?://\S+", "", text)

    # 2. Remove HTML tags
    text = re.sub(r"<[^>]+>", "", text)

    # 3. Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    # 4. Remove leading/trailing whitespace
    text = text.strip()

    # 5. Lowercase
    text = text.lower()

    return text

df["overview_clean"] = df["overview"].apply(preprocess_overview)
```

```python
df[["overview", "overview_clean"]].sample(
    3,
    random_state=42
)
```

```python
# We deliberately left some important punctuation marks that keeps meaning in the sentence .
# Also we are not removing stopwards for now to actual hold current context as it is.
```

```python
df["tokens"] = df["overview_clean"].apply(word_tokenize) # Store tokens in a new column temporarily
```

```python
# Currently we have decidede to not remove this stopwords as this also contain important linguistic patterns especially negation which will help in our multi-genre classifier later.
```

```python
sample_text = df["overview_clean"].sample(
    1,random_state=42
).iloc[0]


tokens = word_tokenize(sample_text)

lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [
    lemmatizer.lemmatize(token)
    for token in tokens
]

for original, lemma in zip(tokens, lemmatized_tokens):
    if original != lemma:
        print(f"{original:20} → {lemma}")
```
